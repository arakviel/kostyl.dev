# 🔒 Службові хелпери

Ці хелпери дає викладач. Вони потрібні, щоб **автоматично перевіряти** ваші програми, які читають дані через `input()`.

Як це працює: замість того щоб програма чекала на справжній ввод, `set_inputs(...)` підставляє заготовлені відповіді — по одному на кожен виклик `input()`.

```python
set_inputs("30", "41")   # перший input() поверне "30", другий — "41"
```

Після виконання блоку з `input()` обов'язково викликайте `reset_inputs()`, щоб повернути справжній `input()`.

````python
# 🔒 Службова клітинка: хелпери для перевірки (не змінюйте)
import builtins

_ORIGINAL_INPUT = builtins.input
_INPUT_QUEUE = []


def _fake_input(prompt=""):
    """Імітує input(): показує підказку і повертає наступне задане значення."""
    print(prompt, end="")
    if not _INPUT_QUEUE:
        raise RuntimeError(
            "❌ Програма попросила більше даних, ніж задано в set_inputs(...)."
        )
    return _INPUT_QUEUE.pop(0)


def set_inputs(*values):
    """Готує відповіді на виклики input() — по порядку."""
    global _INPUT_QUEUE
    _INPUT_QUEUE = [str(v) for v in values]
    builtins.input = _fake_input


def reset_inputs():
    """Повертає справжній input()."""
    builtins.input = _ORIGINAL_INPUT


def _need(*names):
    """Перевіряє, що змінна вже створена, інакше — зрозуміла помилка."""
    missing = [name for name in names if name not in globals()]
    if missing:
        raise AssertionError(
            "❌ Спершу виконайте клітинку із завданням: " + ", ".join(missing)
        )


print("🔒 Хелпери завантажено: set_inputs(...), reset_inputs()")
````

# Блок 1. `print()` — виведення інформації

## Що ми вже знаємо

`print()` — вбудована функція Python, яка показує текст на екрані.

```python
print("Hello, World!")
```

* **Аргументи** — все, що ми передаємо в дужках. Їх може бути багато, і вони будуть розділені пробілом.
* **Позиційні аргументи** — просто перелік значень: `print("a", "b")`.
* **Ключові аргументи** — мають ім'я. Два найважливіші для нас:

| Аргумент | Що робить | За замовчуванням |
| :--- | :--- | :--- |
| `sep` | яким символом розділяти позиційні аргументи | пробіл `" "` |
| `end` | що додати в кінці виводу | новий рядок `"\n"` |

```python
print("My", "name", "is", "Monty", "Python.", sep="-")
# My-name-is-Monty-Python.
```

```python
print("My name is", end=" ")
print("Monty Python.")
# My name is Monty Python.
```

## Як ми перевірятимемо `print()`

Звичайний `assert` не бачить того, що надруковано на екрані. Тому ми «записуємо» вивід у спеціальну змінну за допомогою магічного рядка `%%capture`:

```python
%%capture cap
print("Hello")
```

Після виконання в змінній `cap` лежить об'єкт, у якого `cap.stdout` — це те, що було надруковано. У клітинці перевірки ми порівняємо цей рядок із очікуваним.

> ⚠️ Важливо: рядок `%%capture cap` — перший у клітинці, і його не треба змінювати.

### 👀 Де буде мій вивід?

Магічний рядок `%%capture` **записує** вивід, але **не показує** його в клітинці — інакше ми не змогли б його перевірити. Тому:

* 📋 **Ваш вивід буде показано в клітинці «Перевірка»** (наступна клітинка) — там ви побачите саме те, що надрукував Python, і порівняння з очікуваним.
* 👀 **Хочете бачити вивід одразу під час написання?** Тимчасово закоментуйте рядок `%%capture cap` (додайте `#` у початок) — тоді `print()` показуватиме свій результат як завжди. Коли все готово — поверніть рядок як було, інакше перевірка не спрацює.

````python
# 🔒 Демонстрація (викладач)
print("My", "name", "is", "Monty", "Python.", sep="-")   # роздільник задано вручну
print("My", "name", "is", "Monty", "Python.")            # типовий вивід: пробіли
print("Кінець без переносу:", end="")                     # без переносу рядка
print(" <- тут ми продовжили те саме рядок")
````

# Блок 2. Літерали та типи даних

## Що ми вже знаємо

**Літерал** — це фіксоване значення, записане прямо в коді (`123`, `"Hello"`, `True`).

Основні типи цього уроку:

| Тип | Що це | Приклади |
| :--- | :--- | :--- |
| `int` | ціле число | `10`, `-5`, `0` |
| `float` | дробове число | `3.14`, `-0.4`, `4.0` |
| `str` | текст (рядок) | `"Привіт"`, `'Python'` |
| `bool` | істина/хибність | `True`, `False` |

Функція `type(x)` показує тип значення.

### Інші способи запису чисел

| Запис | Що це | Значення |
| :--- | :--- | :--- |
| `1_000_000` | підкреслення для читабельності | 1000000 |
| `0o123` | вісімкова система | 83 |
| `0x123` | шістнадцяткова система | 291 |
| `3e8` | експоненційний запис | 300000000.0 |

> ⚠️ `4` — це `int`, а `4.0` — це `float`. Наявність точки змінює тип!

### Перетворення типів

`int("30")` → `30`, `float("3.5")` → `3.5`, `str(30)` → `"30"`.

````python
# 🔒 Демонстрація (викладач)
print("type(42)      =", type(42))
print("type(3.14)    =", type(3.14))
print("type('Python')=", type('Python'))
print("type(True)    =", type(True))

print()
print("0o123 =", 0o123, "| 0x123 =", 0x123, "| 3e8 =", 3e8)
print("type(0o123) =", type(0o123), "| type(3e8) =", type(3e8))
print("4   це", type(4), " а 4.0 це", type(4.0))

print()
print("int('30') =", int("30"), type(int("30")))
print("float('3.5') =", float("3.5"), type(float("3.5")))
print("str(30) =", repr(str(30)), type(str(30)))
````

# Блок 3. Рядки: екранування, escape-послідовності, конкатенація

## Що ми вже знаємо

Рядок створюється подвійними або одинарними лапками: `"Hello"`, `'Hello'`, і навіть порожній: `""`.

# Блок 4. Арифметичні оператори та пріоритети

## Що ми вже знаємо

| Оператор | Опис | Приклад | Результат |
| :--- | :--- | :--- | :--- |
| `+` | додавання | `5 + 3` | `8` |
| `-` | віднімання | `5 - 3` | `2` |
| `*` | множення | `5 * 3` | `15` |
| `/` | ділення (завжди `float`!) | `10 / 3` | `3.333…` |
| `//` | цілочисельне ділення | `10 // 3` | `3` |
| `%` | остача від ділення | `10 % 3` | `1` |
| `**` | піднесення до степеня | `2 ** 3` | `8` |

### Пріоритети

1. `**` — піднесення до степеня
2. `*`, `/`, `//`, `%`
3. `+`, `-`

**Зв'язування:** більшість операторів — зліва направо, але `**` — **справа наліво**:
`2 ** 2 ** 3` → `2 ** (2 ** 3)` → `256`.

### Дві важливі особливості

* `//` — округлює **до найближчого меншого цілого** (донизу), а не просто відкидає дробову частину: `-6 // 4` = `-2`.
* `%` — знак результату має той самий знак, що й **ділене**: `-14 % 4` = `2`, `14 % -4` = `-2`.

````python
# 🔒 Демонстрація (викладач)
print("6 / 3   =", 6 / 3, type(6 / 3))      # 2.0 — ділення завжди дає float
print("6 // 4  =", 6 // 4)                   # 1
print("-6 // 4 =", -6 // 4)                  # -2 (округлення вниз!)
print("14 % 4  =", 14 % 4)                   # 2
print("-14 % 4 =", -14 % 4)                  # 2  (знак діленого)
print("14 % -4 =", 14 % -4)                  # -2 (знак діленого)
print("2 ** 2 ** 3 =", 2 ** 2 ** 3)          # 256 (справа наліво)
print("2 * 3 + 4  =", 2 * 3 + 4)             # 10
print("2 * (3 + 4)=", 2 * (3 + 4))           # 14
````

## Що ми вже знаємо

**Змінна** — це іменована «коробка» для значення. Створюється в момент присвоєння `=`.

```python
user_age = 30
print(user_age)
```

### Правила іменування

* ✅ можна: літери `a-z`, `A-Z`, цифри `0-9`, підкреслення `_`;
* ✅ ім'я має починатися з літери або `_`;
* ✅ регістр важливий: `my_variable` ≠ `My_Variable`;
* ❌ не можна: починати з цифри (`1st_place`), пробіли (`exchange rate`);
* ❌ не можна: **зарезервовані слова** (`if`, `else`, `for`, `while`, `True`, `False`, `None`, `import`).

### Скорочені оператори присвоєння

| Повний запис | Скорочений | Зміст |
| :--- | :--- | :--- |
| `x = x + 1` | `x += 1` | збільшити `x` на 1 |
| `x = x - 5` | `x -= 5` | зменшити `x` на 5 |
| `x = x * 2` | `x *= 2` | помножити `x` на 2 |

````python
# 🔒 Демонстрація (викладач)
counter = 1
print("counter =", counter)
counter = counter + 1
print("counter =", counter)
counter += 1
print("counter =", counter)

# Різні змінні — це різні імена
price = 100
Price = 200
print("price =", price, "| Price =", Price)
````

age_int = int(age_str)      # "30"  → 30

````python
# 🔍 ФІНАЛЬНА ПЕРЕВІРКА (викладач)
RESULTS = []


def _record(label, check_fn):
    try:
        check_fn()
        RESULTS.append((label, "✅", "пройдено"))
    except Exception as exc:
        RESULTS.append((label, "❌", str(exc)))


def _check_block_1():
    _need("cap1", "cap2", "cap3")
    assert cap1.stdout == "Python-урок-1\n", "ТЗ-1: неправильний sep"
    assert cap2.stdout == "Ім'я: Олесь\n", "ТЗ-2: неправильний end"
    assert cap3.stdout == "Курс-Python-1!", "ТЗ-3: неправильні sep та end разом"


def _check_block_2():
    _need("num", "ratio", "text", "flag", "hex_num", "oct_num", "sci_num",
          "pred_div", "pred_mul", "pred_concat")
    assert type(num) is int and type(ratio) is float, "ТЗ-4: неправильні типи"
    assert type(text) is str and type(flag) is bool, "ТЗ-4: неправильні типи"
    assert hex_num == 31 and oct_num == 15 and sci_num == 3000.0, "ТЗ-5: неправильні значення"
    assert pred_div == "float" and pred_mul == "int" and pred_concat == "str", "ТЗ-6: неправильний тип"


def _check_block_3():
    _need("quote_line", "card", "full_name", "age_line")
    assert quote_line == 'Він сказав: "Привіт"!', "ТЗ-7: неправильний рядок з лапками"
    assert card == "Ім'я: Олесь\nВік: 30", "ТЗ-8: неправильний перенос рядка"
    assert full_name == "Олесь Ковальчук", "ТЗ-9: неправильна конкатенація"
    assert age_line == "Вік: 30", "ТЗ-10: неправильне перетворення числа на рядок"


def _check_block_4():
    _need("total", "part", "whole", "rest", "power", "priority", "brackets",
          "p_floor", "p_mod", "p_pow", "t_mul", "t_mul_float", "t_pow", "t_pow_float")
    assert total == 22 and part == 3.4 and whole == 3 and rest == 2, "ТЗ-11: неправильні обчислення"
    assert power == 1419857, "ТЗ-11: неправильний степінь"
    assert priority == 14 and brackets == 20, "ТЗ-11: неправильний пріоритет операторів"
    assert p_floor == -4 and p_mod == -1 and p_pow == 512, "ТЗ-12: неправильне передбачення"
    assert t_mul == "int" and t_mul_float == "float", "ТЗ-13: неправильні типи"
    assert t_pow == "int" and t_pow_float == "float", "ТЗ-13: неправильні типи"


def _check_block_5():
    _need("income", "valid_names", "invalid_names", "import_is_reserved")
    assert income == 240, "ТЗ-14: неправильний результат лічильника"
    assert "user_name" in valid_names and "class" not in valid_names, "ТЗ-15: неправильний поділ імен"
    assert import_is_reserved is False, "ТЗ-16: 'import' є зарезервованим словом"


def _check_block_6():
    _need("age_str", "age_int", "next_year_age", "side_str", "side", "square_area",
          "concat_result", "number_result")
    assert age_int == 30 and next_year_age == 31, "ТЗ-17: неправильний вік"
    assert side == 3.5 and square_area == 12.25, "ТЗ-18: неправильна площа квадрата"
    assert concat_result == "55" and number_result == 10, "ТЗ-19: неправильні результати"


def _check_block_7():
    _need("width", "height", "area", "uah", "rate", "usd", "c")
    assert area == 20.0, "ТЗ-20: неправильна площа прямокутника"
    assert abs(usd - 2500 / 41.5) < 0.01, "ТЗ-21: неправильна сума в USD"
    assert c == 5.0, "ТЗ-22: неправильна гіпотенуза"


def _check_block_8():
    _need("b1", "b2", "b3", "b4", "b5", "b6")
    assert b1 == 14 and b2 == 20, "Бліц: неправильний пріоритет"
    assert b3 == 3 and b4 == 1, "Бліц: неправильне ділення"
    assert b5 == "str" and b6 is True, "Бліц: неправильні типи/порівняння"


_record("Блок 1 — print(), sep та end", _check_block_1)
_record("Блок 2 — літерали та типи даних", _check_block_2)
_record("Блок 3 — рядки та екранування", _check_block_3)
_record("Блок 4 — арифметичні оператори", _check_block_4)
_record("Блок 5 — змінні та іменування", _check_block_5)
_record("Блок 6 — input() та перетворення типів", _check_block_6)
_record("Блок 7 — підсумкові задачі", _check_block_7)
_record("Блок 8 — бліц-тест", _check_block_8)

print("=" * 72)
print("ПІДСУМКОВА ПЕРЕВІРКА — Урок 1: Основи Python")
print("=" * 72)
for label, status, message in RESULTS:
    print(f"{status}  {label}")
    if status == "❌":
        print(f"       → {message}")
passed = 0
for _, status, _ in RESULTS:
    if status == "✅":
        passed += 1
print("=" * 72)
print(f"Пройдено блоків: {passed} з {len(RESULTS)}")
print("=" * 72)

if passed == len(RESULTS):
    print("🎉 Чудово! Усі завдання виконано правильно. Практикум засвоєно.")
else:
    print("👉 поверніться до блоків із ❌ і виправте помилки, потім запустіть цю клітинку ще раз.")
````
