# 🔒 Службові хелпери

Ці інструменти необхідні для **автоматичної перевірки** ваших програм, які взаємодіють з користувачем через `input()`.

* `set_inputs("значення_1", "значення_2")` — заздалегідь підставляє тестові дані у чергу для `input()`.
* `reset_inputs()` — скидає підстановку та повертає стандартний `input()`.
* `_need("var1", "var2")` — перевіряє, чи створені змінні у вашому зошиті.

````python
# 🔒 Службова клітинка: хелпери перевірки (не змінюйте)
import builtins

_ORIGINAL_INPUT = builtins.input
_INPUT_QUEUE = []


def _fake_input(prompt=""):
    """Імітує input(): виводить підказку і повертає заготовлене значення."""
    print(prompt, end="")
    if not _INPUT_QUEUE:
        raise RuntimeError(
            "❌ Програма викликала input() більше разів, ніж задано значень у set_inputs(...)."
        )
    return _INPUT_QUEUE.pop(0)


def set_inputs(*values):
    """Задає тестові дані для викликів input()."""
    global _INPUT_QUEUE
    _INPUT_QUEUE = [str(v) for v in values]
    builtins.input = _fake_input


def reset_inputs():
    """Відновлює стандартний системний input()."""
    builtins.input = _ORIGINAL_INPUT


def _need(*names):
    """Перевіряє наявність змінних у пам'яті зошита."""
    missing = [name for name in names if name not in globals()]
    if missing:
        raise AssertionError(
            "❌ Спершу виконайте клітинку із завданням. Відсутні змінні: " + ", ".join(missing)
        )


print("🔒 Хелпери завантажено: set_inputs(), reset_inputs(), _need()")
````

# Блок 1. Числові літерали: підкреслення, системи числення та експонента

## Теорія

1. **Підкреслення в числах (Python 3.6+, PEP 515):**
   Для покращення читабельності великих чисел можна використовувати символ підкреслення `_`. Комп'ютер його просто ігнорує:
   ```python
   world_population = 8_000_000_000  # значно легше читати, ніж 8000000000
   distance_km = 384_400
   ```
   *Зауваження:* не можна ставити `_` на початку числа, в кінці або поруч із крапкою.

2. **Системи числення:**
   За замовчуванням усі числа в Python записуються в десятковій системі числення. Але Python підтримує літерали інших систем:
   * **Двійкова (Binary):** префікс `0b` або `0B` (цифри `0`, `1`). Наприклад, `0b1010` — це число `10`.
   * **Вісімкова (Octal):** префікс `0o` або `0O` (цифри `0..7`). Наприклад, `0o12` — це число `10`.
   * **Шістнадцяткова (Hexadecimal):** префікс `0x` або `0X` (цифри `0..9`, `A..F` або `a..f`). Наприклад, `0xA` — це `10`, `0xFF` — це `255`.

   Вбудовані функції для отримання рядкового представлення у відповідних системах:
   * `bin(42)` → `'0b101010'`
   * `oct(42)` → `'0o52'`
   * `hex(42)` → `'0x2a'`

3. **Експоненційний запис (наукова нотація):**
   Використовується літера `e` або `E` для запису чисел виду $m 	imes 10^k$:
   ```python
   speed_of_light = 3e8     # 3 * 10^8 = 300000000.0 (тип float!)
   planck_const = 6.626e-34 # 6.626 * 10^-34
   ```
   > ⚠️ Будь-яке число, записане з літерою `e` чи `E`, автоматично отримує тип **`float`**, навіть якщо в нього немає десяткової крапки.

````python
# 🔒 Демонстрація викладача: системи числення та науковий запис
val_dec = 255
val_bin = 0b11111111
val_oct = 0o377
val_hex = 0xFF

print("Десяткове:", val_dec)
print("Двійковий літерал:", val_bin, "| Через bin():", bin(val_dec))
print("Вісімковий літерал:", val_oct, "| Через oct():", oct(val_dec))
print("Hex літерал:", val_hex, "| Через hex():", hex(val_dec))
print("Усі значення рівні?", val_dec == val_bin == val_oct == val_hex)

sci = 1.5e3
print("1.5e3 дорівнює:", sci, "| Тип:", type(sci).__name__)
````

# Блок 2. Рядки: екранування, raw-рядки та багаторядковий текст

## Теорія

1. **Лапки та екранування:**
   У Python рядки можна огортати як одинарними `'...'`, так і подвійними `"..."` лапками.
   Якщо всередині рядка потрібні лапки того ж типу, що й зовнішні, їх необхідно екранувати символом `\`:
   ```python
   quote = "Він сказав: \"Привіт!\""
   ```
   Найважливіші escape-послідовності:
   * `\n` — перенос на новий рядок
   * `\t` — горизонтальна табуляція
   * `\\` — символ зворотної косої риски

2. **Сирі рядки (Raw strings):**
   Якщо перед лапками додати префікс `r` або `R`, Python відключає інтерпретацію escape-послідовностей. Символи `\` залишаються звичайними символами.
   Це незамінно для шляхів до файлів у Windows та регулярних виразів:
   ```python
   # Без r: \n перетвориться на новий рядок!
   bad_path = "C:\new_folder\test" 
   
   # З r: залишається чистим шляхом
   good_path = r"C:\new_folder\test"
   ```

3. **Багаторядкові рядки (Multiline strings):**
   Потрійні лапки `"""..."""` або `'''...'''` дозволяють писати текст на багатьох рядках, автоматично зберігаючи всі переноси рядків та відступи без необхідності писати `\n`.

````python
# 🔒 Демонстрація викладача: escape-послідовності, raw-рядки та потрійні лапки
escaped = "рядок 1\nрядок 2\tз табуляцією"
raw = r"C:\tools\new_project\read.txt"
multiline = """Перший рядок
Другий рядок
Третій рядок"""

print("--- Escaped ---")
print(escaped)
print("--- Raw ---")
print(raw)
print("--- Multiline ---")
print(multiline)
````

# Блок 3. Float-точність, булеві значення та None

## Теорія

1. **Проблема точності чисел з рухомою крапкою (`float`):**
   Комп'ютери зберігають числа у двійковій системі за стандартом IEEE 754. Число `0.1` у двійковій системі — це нескінченний періодичний дріб (як `1/3 = 0.3333...` у десятковій).
   Тому:
   ```python
   0.1 + 0.2  # повертає 0.30000000000000004
   0.1 + 0.2 == 0.3  # повертає False!
   ```
   **Рішення:** Для порівняння або виведення використовують функцію округлення `round(value, n_digits)`:
   ```python
   round(0.1 + 0.2, 2) == 0.3  # повертає True
   ```

2. **Булевий тип як підтип цілих чисел (`bool`):**
   В Python тип `bool` історично успадковує `int` (`issubclass(bool, int) is True`).
   * `True` чисельно дорівнює `1`
   * `False` чисельно дорівнює `0`
   * Їх можна використовувати в арифметиці: `True + True == 2`, `False * 10 == 0`.

3. **Спеціальне значення `None`:**
   `None` представляє відсутність значення і має власний тип `NoneType`.
   * Перевірку на `None` прийнято виконувати за допомогою оператора ідентичності **`is`** (`val is None`, `val is not None`), а не `==`.
   * `None` не дорівнює `0`, `False` чи порожньому рядку `""`. Вони мають абсолютно різні типи даних!

4. **Таблиця істинності (Truthiness):**
   Будь-який об'єкт у Python можна перевірити на логічне значення функцією `bool()`:
   * **Хибні (Falsy):** `0`, `0.0`, `""` (порожній рядок), `None`, `False`.
   * **Істинні (Truthy):** будь-яке ненульове число (`1`, `-5`, `0.001`), непорожній рядок (`"Hello"`, `" "`).

````python
# 🔒 Демонстрація викладача: точність float, bool та None
print("0.1 + 0.2 =", 0.1 + 0.2)
print("0.1 + 0.2 == 0.3 ->", 0.1 + 0.2 == 0.3)
print("round(0.1 + 0.2, 2) == 0.3 ->", round(0.1 + 0.2, 2) == 0.3)

print("True + True =", True + True)
print("None is None ->", None is None)
print("None == 0 ->", None == 0)
print("bool(''), bool('a') ->", bool(""), bool("a"))
````

   * Для звичайних змінних офіційний стандарт Python (PEP 8) вимагає стиль **`snake_case`**: слова пишуться маленькими літерами і розділяються підкресленням: `user_name`, `total_score`.

   У чистому Python немає захищених констант на рівні мови, але розробники домовилися записувати змінні, які не повинні змінюватися, великими літерами з підкресленнями:
   ```python
   PI = 3.14159
   MAX_RETRIES = 5
   ```

3. **Множинне та паралельне присвоєння:**
   * Одне значення кільком змінним:

     ```python
     a = b = c = 0
     ```
   * Різні значення кільком змінним одночасно (кортежне розпакування):
     ```python
     x, y, z = 10, 20, 30
     ```

4. **Pythonic обмін значеннями двох змінних:**
   В інших мовах (наприклад, C чи Pascal) для обміну значень потрібна третя тимчасова змінна. У Python це робиться елегантно в один рядок:
   ```python
   a, b = b, a
   ```

# Блок 5. Скорочені оператори, динамічна типізація та ключові слова

## Теорія

1. **Скорочені оператори (Augmented Assignment):**
   Замість `x = x + 5` у Python прийнято писати скорочено:
   * `x += 5` — додавання
   * `x -= 3` — віднімання
   * `x *= 2` — множення
   * `x /= 4` — ділення
   * `x //= 2` — цілочисельне ділення
   * `x %= 5` — залишок від ділення
   * `x **= 3` — піднесення до степеня

2. **Динамічна типізація:**
   У Python тип даних прив'язаний не до назви змінної, а до самого об'єкта. Одна й та сама змінна може в різні моменти часу містити число, рядок чи булеве значення:
   ```python
   data = 100       # data вказує на int
   data = "Привіт"   # тепер data вказує на str
   ```
   * `type(data)` — повертає тип об'єкта
   * `isinstance(data, str)` — перевіряє, чи належить об'єкт заданому типу

3. **Ключові слова (Keywords):**
   Ключові слова — це зарезервовані мовою слова (`def`, `class`, `import`, `for`, `while` тощо), які заборонено використовувати як імена змінних.
   Модуль `keyword` містить список усіх зарезервованих слів:
   ```python
   import keyword
   print(keyword.kwlist)               # повний список
   print(keyword.iskeyword("import"))  # True
   ```

````python
# 🔒 Демонстрація викладача: скорочені оператори та перевірка типу
counter = 10
counter += 5   # 15
counter *= 2   # 30
counter //= 4  # 7
print("Кінцевий counter:", counter)

val = 42
print(f"val = {val}, isinstance(val, int):", isinstance(val, int))
val = "Text"
print(f"val = {val}, isinstance(val, str):", isinstance(val, str))
````

# Блок 6. Перетворення систем числення, операції з рядками та f-strings

## Теорія

1. **Функція `int(string, base)`:**
   Функція `int()` вміє не лише перетворювати рядок на десяткове число, але й зчитувати числа з будь-якої системи числення (від 2 до 36), якщо вказати аргумент `base`:
   ```python
   int("1010", 2)   # 10 (двійкова система)
   int("77", 8)     # 63 (вісімкова система)
   int("FF", 16)    # 255 (шістнадцяткова система)
   ```

2. **Операції з рядками:**
   * **Конкатенація (`+`):** об'єднання рядків. Працює лише між двома рядками: `"Hello" + " " + "World"`. Якщо додати число, виникне помилка `TypeError`. Число слід явно обернути в `str()`.
   * **Реплікація (`*`):** повторення рядка задану кількість разів. Наприклад:
     ```python
     "ha" * 3       # 'hahaha'
     "-" * 20       # '--------------------'
     ```
     Це дуже зручно для малювання ASCII-рамок, таблиць або розділювачів.

3. **Форматування f-strings (Python 3.6+):**
   Усередині f-рядка `{значення:специфікатор}` можна форматувати представлення чисел:
   * `{x:b}` — вивести число `x` у двійковому форматі
   * `{x:o}` — вивести у вісімковому форматі
   * `{x:x}` або `{x:X}` — вивести у hex форматі (малими чи великими літерами)
   * `{pi:.2f}` — вивести число з рухомою крапкою рівно з 2 знаками після крапки

````python
# 🔒 Демонстрація викладача: int(..., base), реплікація та f-strings
print("Двійковий 1101 в десятковій:", int("1101", 2))
print("Hex 1F в десятковій:", int("1F", 16))

border = "+" + "=" * 20 + "+"
print(border)
print("|  ФОРМАТУВАННЯ  |")
print(border)

num = 255
pi_val = 3.14159265
print(f"Число {num}: двійкове={num:b}, hex={num:X}")
print(f"Число Пі з 2 знаками: {pi_val:.2f}, з 4 знаками: {pi_val:.4f}")
````
